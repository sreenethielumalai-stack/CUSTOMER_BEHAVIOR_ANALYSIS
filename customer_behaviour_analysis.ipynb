```python
import pandas as pd
import os
from google.colab import files

target_file = "customer_shopping_cleaned.csv"

# If the file does not exist, ask the user to upload it
if not os.path.exists(target_file):
    print(f"'{target_file}' not found in the current session.")
    print("Please upload 'customer_shopping_cleaned.csv' below:")
    uploaded = files.upload()

    # Handle case where file was renamed during upload (e.g. if uploaded multiple times)
    uploaded_name = next(iter(uploaded))
    if uploaded_name != target_file:
        df = pd.read_csv(uploaded_name)
        df.to_csv(target_file, index=False)

# Ensure the file is ready and download it
if os.path.exists(target_file):
    print(f"Downloading {target_file}...")
    files.download(target_file)
else:
    print("File upload was not completed.")
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['young adult','Adult','Middle-aged','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365,
    'Every 3 Months': 90

}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
